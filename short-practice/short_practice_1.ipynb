```python
import numpy as np

import matplotlib.pyplot as plt

# Define x and y
x = np.linspace(0, np.pi, 1001)
y = np.sin(x)

# Plot y(x)
plt.figure(figsize=(8, 4))
plt.plot(x, y, label=r"$y=\sin(x)$")
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Plot of $y=\sin(x)$")
plt.grid(True)
plt.legend()
plt.show()

# Numerical integration using the trapezoidal rule
I_numerical = np.trapezoid(y, x)
I_exact = 2.0
absolute_error = abs(I_numerical - I_exact)

print(f"Numerical integral: {I_numerical:.12f}")
print(f"Exact integral:     {I_exact:.12f}")
print(f"Absolute error:     {absolute_error:.12e}")
```
