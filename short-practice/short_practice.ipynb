# Short Practice: Numerical Integration of sin(x) on [0, pi]

Define y = sin(x) on [0, pi], plot it, integrate numerically, and compare against the exact value: integral of sin(x) dx from 0 to pi = 2.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson
```

## Define x and y

```python
x = np.linspace(0, np.pi, 1000)
y = np.sin(x)
```

## Plot y(x)

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, color='C0')
ax.set_xlabel('x')
ax.set_ylabel('y = sin(x)')
ax.set_title('y = sin(x) on [0, pi]')
ax.grid(True, alpha=0.3)
plt.show()
```

## Numerical integration and comparison with exact value

```python
I_numerical = simpson(y, x=x)
I_exact = 2.0
abs_error = abs(I_numerical - I_exact)

print(f"Numerical integral (Simpson's rule): {I_numerical:.10f}")
print(f"Exact value: {I_exact:.10f}")
print(f"Absolute error |I_numerical - 2|: {abs_error:.2e}")
```
