## Practice 1
Replace sin(x) with another smooth function.

```python
def grid_calculus_general(func, dfunc, exact_integral, a, b, number_of_points):
    x = np.linspace(a, b, number_of_points)
    f = func(x)
    derivative = np.gradient(f, x)
    integral = np.trapezoid(f, x)

    interior = slice(1, -1)
    derivative_error = np.max(np.abs(derivative[interior] - dfunc(x[interior])))
    integral_error = abs(integral - exact_integral)

    return x, f, derivative, integral, derivative_error, integral_error

# cos(x)로 교체: 도함수는 -sin(x), [0, π] 위 적분의 정확값은 sin(π)-sin(0)=0
exact_int_cos = np.sin(np.pi) - np.sin(0.0)
x, f, deriv, integ, d_err, i_err = grid_calculus_general(
    np.cos, lambda x: -np.sin(x), exact_int_cos, 0.0, np.pi, 21
)
print("integral:", integ, " integral error:", i_err)
print("max interior derivative error:", d_err)
```

## Practice 2
Change the number of sample points used for interpolation.

```python
x_query = np.linspace(0.0, 2.0 * np.pi, 2001)
exact_query = np.sin(x_query)

print(f"{'n_samples':>10} {'max interp error':>20}")
for n in [5, 9, 17, 33, 65]:
    x_sample = np.linspace(0.0, 2.0 * np.pi, n)
    f_sample = np.sin(x_sample)
    f_interp = np.interp(x_query, x_sample, f_sample)
    err = np.max(np.abs(f_interp - exact_query))
    print(f"{n:10d} {err:20.6e}")
```

## Practice 3
Find a different root with bisection.

```python
def bisection(func, left, right, tol=1.0e-12, max_iter=100):
    f_left = func(left)
    f_right = func(right)

    if f_left * f_right > 0:
        raise ValueError("The interval does not bracket a root.")

    for iteration in range(1, max_iter + 1):
        middle = 0.5 * (left + right)
        f_middle = func(middle)

        if abs(f_middle) < tol or 0.5 * (right - left) < tol:
            return middle, iteration

        if f_left * f_middle <= 0:
            right = middle
        else:
            left = middle
            f_left = f_middle

    raise RuntimeError("Bisection did not converge.")

cubic = lambda x: x**3 - x - 2.0
root, iterations = bisection(cubic, 1.0, 2.0)

print("root:", root)
print("iterations:", iterations)
print("residual:", cubic(root))

assert abs(cubic(root)) < 1.0e-12
```
